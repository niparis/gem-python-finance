## Cell 1 - 1,000 received in 5 years, at 7% a year

Three things to know before your first line of code:
- `rate = 0.07` gives the name `rate` to the value 0.07. This is a **variable**. Here `=` means "store this", not "equals" as in maths.
- Text after `#` is a **comment**. Python skips it: it is there for you.
- `print(...)` shows a value below the cell.

**Your prediction:** will the present value be more or less than 1,000? Roughly how much? Write your guess here: ...

```python
cash_flow = 1000   # money received in the future
rate = 0.07        # 7% a year, written as a decimal
years = 5

pv = cash_flow / (1 + rate) ** years   # ** means "to the power of"
print(pv)
```

## Cell 2 - Make the output readable

Same number, shown like in a report. `f"..."` is an **f-string**: text with values inside. Python replaces anything in `{ }` with its value. After the `:` you choose the format: `,` adds a thousands separator, `.2f` keeps 2 decimals.

**Your prediction:** how will the number look? Write it here: ...

```python
print(f"PV = {pv:,.2f}")
```

## Cell 3 - A reusable formula

`def` creates a **function**: a formula with a name, like a formula you could reuse in any Excel cell. It takes inputs and gives back (`return`) a result.

The line under `def` is **indented** (moved right by 4 spaces): that is how Python knows it belongs to the function. To use the function, write its name and the inputs in brackets, in the same order: `present_value(1000, 0.07, 5)`.

**Your prediction:** what does `present_value(1000, 0.07, 5)` give? (Look at cell 2.)

```python
def present_value(cash_flow, rate, years):
    return cash_flow / (1 + rate) ** years


print(f"{present_value(1000, 0.07, 5):,.2f}")
```

## Cell 4 - Future value, and the first check

Future value is the opposite move: compound instead of discount. The result of one function can go into another: here the PV goes into `future_value`. You can also name the inputs when you use a function (`rate=0.07`): it is easier to read.

**Check (checks card, entry 1: put the answer back in):** if you take the PV and grow it for 5 years at 7%, you must get 1,000 back.

**Your prediction:** exactly 1,000?

```python
def future_value(cash_flow, rate, years):
    return cash_flow * (1 + rate) ** years


pv = present_value(1000, 0.07, 5)
back = future_value(pv, rate=0.07, years=5)
print(f"PV = {pv:,.2f}")
print(f"FV of that PV = {back:,.2f}")
```

## Cell 5 - A project with several cash flows

A list holds several values in order. **Positions start at 0**, and year 0 (today) is position 0.
So 5 cash flows are 5 dates: years 0 to 4, positions 0 to 4. (This was the bug in Session 0: a 5-year deal has 6 dates, years 0 to 5, but the code made only 5 slots.)

| Year | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Cash flow | -5,000 | 1,000 | 1,000 | 3,000 | 3,000 |
| Position | `[0]` | `[1]` | `[2]` | `[3]` | `[4]` |

**Your prediction:** what do `cash_flows[0]` and `cash_flows[-1]` give?

```python
cash_flows = [-5000, 1000, 1000, 3000, 3000]

print(cash_flows[0])    # first position = year 0
print(cash_flows[-1])   # -1 means: the last one
print(len(cash_flows))  # how many items
```

## Cell 6 - Net present value

NPV = the present value of every cash flow added together. Year 0 is not discounted (`** 0` is 1), year 1 is discounted once, and so on.

`for t, cf in enumerate(cash_flows):` is a **loop**: it goes through the list, one cash flow at a time. `t` is the position (= the year), `cf` is the cash flow at that position.

The indented line runs once for every cash flow. `npv = npv + ...` means: take the current `npv`, add this cash flow's PV, and store the result back in `npv`. It starts at 0 and grows at each turn, like a running total in Excel.

You should get **2,324.36**.

```python
rate = 0.03
npv = 0

for t, cf in enumerate(cash_flows):
    npv = npv + cf / (1 + rate) ** t   # add the PV of this cash flow

print(f"NPV = {npv:,.2f}")
```

## Cell 7 - Invest or not?

`if / else` lets the code decide. Rule: invest if the NPV is above 0. As with `def` and `for`, the line ends with a colon `:` and the lines that belong to it are indented.

```python
if npv > 0:
    print("Invest: the project creates value.")
else:
    print("Do not invest: the project destroys value.")
```

## Cell 8 - Check against Excel

Python says 2,324.36. In Excel, `=NPV(3%, B2:B6)` on the same five cash flows gives **2,256.66**.

**Your prediction:** why are they different?

The answer: **Excel's NPV treats the first value as year 1**, not year 0. It discounts every cash flow one year too much. The fix in Excel is `=B2 + NPV(3%, B3:B6)`: the year 0 cash flow stays outside.

(Checks card, entry 2: compare with Excel, and know its conventions.)

The cell below copies Excel's convention in Python (`start=1`: count from 1) to prove it.

```python
def excel_npv(values, rate):
    total = 0
    for t, v in enumerate(values, start=1):   # Excel: the first value is year 1
        total = total + v / (1 + rate) ** t
    return total


print(f"Excel-style NPV, all five:      {excel_npv(cash_flows, 0.03):,.2f}")

# The fix: year 0 outside, the other four through Excel's NPV
other_four = [1000, 1000, 3000, 3000]
print(f"Year 0 + Excel NPV of the rest: {cash_flows[0] + excel_npv(other_four, 0.03):,.2f}")
```

## Cell 9 - Your first error message

**This cell fails on purpose.** Read the message: **last line first, then the arrow.**

1. The last line says the type of error and what went wrong.
2. The arrow `---->` shows the line that failed.
3. Ask: which value is not what I think it is?

Here the rate comes from Excel as the text `"7%"`, not the number 0.07.

**Your prediction:** what will the last line say?

(Stuck? Ask Gemini to *explain* the error. Then fix it yourself.)

```python
rate_from_excel = "7%"   # text, not a number

pv = 1000 / (1 + rate_from_excel) ** 5
print(pv)
```

The last line: `TypeError: unsupported operand type(s) for +: 'int' and 'str'`. You cannot add a number (`int`) and text (`str`). The arrow points to the line with `1 + rate_from_excel`.

The fix: turn the text into a decimal. Rates are decimals in this course: 7% is `0.07`.

```python
rate = float(rate_from_excel.replace("%", "")) / 100   # "7%" -> "7" -> 7.0 -> 0.07
print(rate)

pv = 1000 / (1 + rate) ** 5
print(f"PV = {pv:,.2f}")
```

## What you learned

**Python**
- **Variable**: a named value (`rate = 0.07`). **Operator `**`**: to the power of. **f-string**: readable output.
- **Function**: `def ... return`, a reusable formula. Functions can call other functions.
- **List**: values in order, positions from 0. **For loop** with `enumerate`: every item, with its position.
- **Condition** (`if / else`) to decide. **Dictionary** to hold the assumptions.
- **Error message**: read the last line first (the error type, here `TypeError`), then the arrow.

**Finance**
- Discount with `/ (1 + rate) ** years`, compound with `* (1 + rate) ** years`. Rates are decimals.
- Year 0 is position 0. N cash flows means N dates.
- Excel's `NPV` treats the first value as year 1. Python and Excel can both be "right" and still disagree.
