## Step 1 - Inputs and two tools

The inputs are in a dictionary. `terminal_value` and `discount` are the functions you built in 1b, ready to use.

```python
inputs = {
    "sales_year1": 300,
    "sales_growth": [0.15, 0.15, 0.10],   # growth in years 2, 3, 4
    "margin": 0.17,
    "interest": 10,
    "tax": 0.30,
    "payout": 0.40,
    "shares": 10,       # millions
    "g_cut": 0.02,      # long-run growth = last growth minus this
    "r": 0.13,
}


def terminal_value(last_div, g, r):
    if g >= r:
        raise ValueError("Growth must be below the required return (g < r).")
    return last_div * (1 + g) / (r - g)


def discount(cash_flows, r):
    total = 0
    for t, cf in enumerate(cash_flows, start=1):
        total = total + cf / (1 + r) ** t
    return total
```
